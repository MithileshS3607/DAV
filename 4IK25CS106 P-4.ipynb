```python
import pandas as pd
import numpy as np
```

```python


data = {
    "Student" : ["A","B","C","D","E"],
    "Gender" : ["Male","Female","Male","Female","Male"],
    "StudyHours" : [5,6,np.nan,4,7],
    "Attendance" : [80,85,75,np.nan,90],
    "Marks" : [75,82,68,5,np.nan],
    "Result" : ["PASS","PASS","PASS","FAIL","PASS"],
}
```

```python
df = pd.DataFrame(data)

display(df)
```

```python
print(df.isnull().sum())
```

```python
df["StudyHours"] = df["StudyHours"].fillna(
    df["StudyHours"].mean()
)
display(df)
```

```python
df["Marks"] = df["Marks"].fillna(
    df["Marks"].median()
)
display(df)
```

```python
df["Attendance"] = df["Attendance"].ffill()
display(df)
```

```python
df["Gender"] = df["Gender"].astype(str).str.strip().str.title()

df["Gender"] = df["Gender"].map({
    "Male": 1,
    "Female": 0
})

display(df)
```
